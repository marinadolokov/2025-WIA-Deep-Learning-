```python
import os
import pickle
import random
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from sklearn.model_selection import KFold
from sklearn.metrics import classification_report, confusion_matrix
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------ Configuration ------------------
# Paths
TRAIN_PATH = '../data/train.pickle'
TEST_PATH = '../data/test.pickle'
MODEL_SAVE_PREFIX = 'brake_diagnosis_2dcnn_binary_'
META_SAVE_PATH = 'global_meta_binary.pkl'

# Random seed & device
SEED = 42

def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(SEED)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using device: {device}')

# Hyperparameters
TRAIN_BATCH_SIZE = 32
EVAL_BATCH_SIZE = 32
EPOCHS = 100
LEARNING_RATE = 3e-3
WEIGHT_DECAY = 1e-4
DROPOUT_RATE = 0.3
NUM_FOLDS = 5
LR_PATIENCE = 5         # for ReduceLROnPlateau
ES_PATIENCE = 150        # stop (EarlyStopping),just for easily adjusting
NUM_WORKERS = 4         # DataLoader workers
PIN_MEMORY = True       # DataLoader pin_memory

# ------------------ Data Processing Functions ------------------
def load_and_prepare_data(file_path):
    """Load pickle file and extract sensor data and binary labels"""
    with open(file_path, 'rb') as f:
        data = pickle.load(f)
    sensor_data = data['sensor_data'].tolist()
    labels = data['label'].tolist()
    # Convert to binary: 0 -> 0 (no anomaly), else 1 (anomaly)
    binary_labels = [0 if label == 0 else 1 for label in labels]
    sample_shape = sensor_data[0].shape
    print(f"Loaded {len(sensor_data)} samples, each with shape {sample_shape}")
    label_counts = Counter(binary_labels)
    print("\nBinary label distribution:")
    for label, count in sorted(label_counts.items()):
        print(f"Class {label}: {count} samples ({count/len(binary_labels)*100:.2f}%)")
    return sensor_data, binary_labels


def normalize_sensor_data(sensor_list, means=None, stds=None):
    """Normalize a list of sensor data arrays. If means/stds not provided, compute globally."""
    sensor_array = np.stack(sensor_list, axis=0)
    if means is None or stds is None:
        means = np.nanmean(sensor_array, axis=(0, 1))
        stds = np.nanstd(sensor_array, axis=(0, 1))
        stds[stds < 1e-8] = 1.0
    normed = (sensor_array - means) / stds
    normed_list = [normed[i] for i in range(normed.shape[0])]
    return normed_list, means, stds

# ------------------ Dataset ------------------
class BrakeDiagnosisDataset(Dataset):
    """Brake diagnosis dataset for binary classification"""
    def __init__(self, sensor_list, label_list):
        assert len(sensor_list) == len(label_list)
        self.sensor_list = sensor_list
        self.label_list = np.array(label_list, dtype=np.int64)
    def __len__(self):
        return len(self.sensor_list)
    def __getitem__(self, idx):
        arr = self.sensor_list[idx]
        x = torch.from_numpy(arr.astype(np.float32)).unsqueeze(0)  # shape: (1, T, C)
        y = int(self.label_list[idx])
        return x, y

# ------------------ Model ------------------
class BrakeDiagnosisCNN(nn.Module):
    """2D CNN model for binary brake anomaly detection"""
    def __init__(self, num_classes=2, dropout_rate=DROPOUT_RATE):
        super(BrakeDiagnosisCNN, self).__init__()
        self.conv_block1 = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=(3, 3), padding=(1, 1)),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2, 1)),
            nn.Dropout(dropout_rate)
        )
        self.conv_block2 = nn.Sequential(
            nn.Conv2d(32, 64, kernel_size=(3, 3), padding=(1, 1)),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2, 1)),
            nn.Dropout(dropout_rate)
        )
        self.conv_block3 = nn.Sequential(
            nn.Conv2d(64, 128, kernel_size=(3, 3), padding=(1, 1)),
            nn.BatchNorm2d(128),
            nn.ReLU()
        )
        self.global_avg_pool = nn.AdaptiveAvgPool2d((1, 1))
        self.classifier = nn.Linear(128, num_classes)
    def forward(self, x):
        x = self.conv_block1(x)
        x = self.conv_block2(x)
        x = self.conv_block3(x)
        x = self.global_avg_pool(x)
        x = x.view(x.size(0), -1)
        return self.classifier(x)

# ------------------ Training with EarlyStopping ------------------
class EarlyStopping:
    """Early stops the training if validation loss doesn't improve after a given patience."""
    def __init__(self, patience=ES_PATIENCE):
        self.patience = patience
        self.counter = 0
        self.best_loss = None
        self.early_stop = False
    def __call__(self, val_loss):
        if self.best_loss is None or val_loss < self.best_loss:
            self.best_loss = val_loss
            self.counter = 0
        else:
            self.counter += 1
            if self.counter >= self.patience:
                self.early_stop = True


def train_model(X_train_normed, y_train, X_val_normed, y_val,
                batch_size=TRAIN_BATCH_SIZE, epochs=EPOCHS, lr=LEARNING_RATE,
                weight_decay=WEIGHT_DECAY):
    """Train the brake diagnosis model with given normalized data."""
    train_dataset = BrakeDiagnosisDataset(X_train_normed, y_train)
    val_dataset = BrakeDiagnosisDataset(X_val_normed, y_val)
    # Handle class imbalance
    class_counts = np.bincount(y_train)
    class_weights = 1.0 / (class_counts + 1e-6)
    sample_weights = class_weights[y_train]
    sampler = WeightedRandomSampler(
        weights=torch.from_numpy(sample_weights).float(),
        num_samples=len(sample_weights), replacement=True
    )
    train_loader = DataLoader(
        train_dataset, batch_size=batch_size,
        sampler=sampler, num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY
    )
    val_loader = DataLoader(
        val_dataset, batch_size=batch_size,
        shuffle=False, num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY
    )
    model = BrakeDiagnosisCNN(num_classes=2, dropout_rate=DROPOUT_RATE).to(device)
    optimizer = optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
    criterion = nn.CrossEntropyLoss()
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode='min', factor=0.5, patience=LR_PATIENCE
    )
    early_stopper = EarlyStopping(patience=ES_PATIENCE)
    best_val_loss = float('inf')
    best_model_wts = None
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': [], 'lr': []}

    for epoch in range(1, epochs + 1):
        model.train()
        train_loss, correct, total = 0.0, 0, 0
        for inputs, targets in train_loader:
            inputs, targets = inputs.to(device), targets.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            loss.backward()
            optimizer.step()
            train_loss += loss.item() * inputs.size(0)
            _, predicted = outputs.max(1)
            correct += predicted.eq(targets).sum().item()
            total += targets.size(0)
        train_loss /= total
        train_acc = correct / total
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        # Validation
        model.eval()
        val_loss, val_correct, val_total = 0.0, 0, 0
        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs, targets = inputs.to(device), targets.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, targets)
                val_loss += loss.item() * inputs.size(0)
                _, predicted = outputs.max(1)
                val_correct += predicted.eq(targets).sum().item()
                val_total += targets.size(0)
        val_loss /= val_total
        val_acc = val_correct / val_total
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        # Record current lr
        current_lr = optimizer.param_groups[0]['lr']
        history['lr'].append(current_lr)
        # Scheduler step
        scheduler.step(val_loss)
        # Early stopping check
        early_stopper(val_loss)
        print(f"Epoch {epoch}/{epochs}: Train Loss: {train_loss:.4f}, Acc: {train_acc:.4f} | "
              f"Val Loss: {val_loss:.4f}, Acc: {val_acc:.4f}, LR: {current_lr:.6f}")
        # Best model
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_model_wts = model.state_dict().copy()
            best_epoch = epoch
            print(f"New best model at epoch {epoch} with val loss: {val_loss:.4f}")
        # Early stop
        if early_stopper.early_stop:
            print(f"Early stopping at epoch {epoch}")
            break
    # Load best weights
    if best_model_wts is not None:
        model.load_state_dict(best_model_wts)
    # Attach best_epoch info
    history['best_epoch'] = best_epoch if 'best_epoch' in locals() else None
    return model, history

# ------------------ Evaluation ------------------
def evaluate_model(model, sensor_data, labels, means, stds,
                   dataset_name="Test", silent=False,
                   batch_size=EVAL_BATCH_SIZE):
    """Evaluate model performance using global normalization parameters"""
    normed_data, _, _ = normalize_sensor_data(sensor_data, means, stds)
    dataset = BrakeDiagnosisDataset(normed_data, labels)
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=False,
                        num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY)
    model.eval()
    all_preds, all_probs, all_targets = [], [], []
    with torch.no_grad():
        for inputs, targets in loader:
            inputs = inputs.to(device)
            outputs = model(inputs)
            probs = torch.softmax(outputs, dim=1)
            _, preds = torch.max(outputs, 1)
            all_preds.extend(preds.cpu().numpy())
            all_probs.append(probs.cpu().numpy())
            all_targets.extend(targets.numpy())
    all_probs = np.concatenate(all_probs, axis=0)
    accuracy = np.mean(np.array(all_preds) == np.array(all_targets))
    if not silent:
        print(f"\n{dataset_name} Accuracy: {accuracy:.4f}")
        print(f"\n{classification_report(all_targets, all_preds, zero_division=0, target_names=['No Irregularity', 'Irregularity'])}")
        cm = confusion_matrix(all_targets, all_preds)
        plt.figure(figsize=(8, 6))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                    xticklabels=['No Irregularity', 'Irregularity'], 
                    yticklabels=['No Irregularity', 'Irregularity'])
        plt.title(f'{dataset_name} Confusion Matrix')
        plt.xlabel('Predicted Label')
        plt.ylabel('True Label')
        plt.savefig(f'{dataset_name.lower().replace(" ", "_")}_confusion_matrix.png')
        plt.show()
    return all_preds, all_probs, all_targets

# ------------------ Visualization ------------------
def plot_training_history(history, fold_idx=None):
    """Plot training and validation loss/accuracy and learning rate"""
    plt.figure(figsize=(15, 6))
    plt.subplot(1, 3, 1)
    plt.plot(history['train_loss'], label='Train Loss')
    plt.plot(history['val_loss'], label='Validation Loss')
    title = 'Training and Validation Loss'
    if fold_idx is not None:
        title = f'Fold {fold_idx+1} - ' + title
    plt.title(title)
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    plt.subplot(1, 3, 2)
    plt.plot(history['train_acc'], label='Train Accuracy')
    plt.plot(history['val_acc'], label='Validation Accuracy')
    title = 'Training and Validation Accuracy'
    if fold_idx is not None:
        title = f'Fold {fold_idx+1} - ' + title
    plt.title(title)
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.subplot(1, 3, 3)
    if 'lr' in history:
        plt.plot(history['lr'], label='Learning Rate')
        title = 'Learning Rate'
        if fold_idx is not None:
            title = f'Fold {fold_idx+1} - ' + title
        plt.title(title)
        plt.xlabel('Epochs')
        plt.ylabel('LR')
        plt.legend()
    plt.tight_layout()
    filename = 'training_history.png' if fold_idx is None else f'training_history_fold{fold_idx+1}.png'
    plt.savefig(filename)
    plt.show()

# ------------------ Ensemble Predictions ------------------
def soft_voting_predict(models, sensor_data, means, stds, batch_size=EVAL_BATCH_SIZE):
    """Soft voting ensemble prediction (average probabilities)"""
    normed_data, _, _ = normalize_sensor_data(sensor_data, means, stds)
    dataset = BrakeDiagnosisDataset(normed_data, [0] * len(normed_data))
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=False,
                        num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY)
    all_probs = []
    for model in models:
        model.eval()
        probs = []
        with torch.no_grad():
            for inputs, _ in loader:
                inputs = inputs.to(device)
                outputs = model(inputs)
                probs_batch = torch.softmax(outputs, dim=1).cpu().numpy()
                probs.append(probs_batch)
        all_probs.append(np.concatenate(probs, axis=0))
    avg_probs = np.mean(all_probs, axis=0)
    predictions = np.argmax(avg_probs, axis=1)
    return predictions, avg_probs

def hard_voting_predict(models, sensor_data, means, stds, batch_size=EVAL_BATCH_SIZE):
    """Hard voting ensemble prediction (majority vote)"""
    normed_data, _, _ = normalize_sensor_data(sensor_data, means, stds)
    dataset = BrakeDiagnosisDataset(normed_data, [0] * len(normed_data))
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=False,
                        num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY)
    all_preds = []
    for model in models:
        model.eval()
        preds = []
        with torch.no_grad():
            for inputs, _ in loader:
                inputs = inputs.to(device)
                outputs = model(inputs)
                _, batch_preds = torch.max(outputs, 1)
                preds.extend(batch_preds.cpu().numpy())
        all_preds.append(preds)
    all_preds = np.array(all_preds)
    final_preds = []
    for i in range(all_preds.shape[1]):
        votes = all_preds[:, i]
        counts = np.bincount(votes)
        final_preds.append(np.argmax(counts))
    return np.array(final_preds)

# ------------------ Main Workflow ------------------
if __name__ == '__main__':
    print("Starting brake diagnosis model training with binary classification...")
    # 1. Load data
    train_sensor_raw, train_labels = load_and_prepare_data(TRAIN_PATH)
    test_sensor_raw, test_labels = load_and_prepare_data(TEST_PATH)
    # 2. Compute global normalization parameters
    print("\nCalculating global normalization parameters from training data...")
    _, global_means, global_stds = normalize_sensor_data(train_sensor_raw)
    print("Global normalization parameters calculated.")
    # 3. Normalize data
    print("\nNormalizing data using global parameters...")
    train_sensor_normed, _, _ = normalize_sensor_data(train_sensor_raw, global_means, global_stds)
    test_sensor_normed, _, _ = normalize_sensor_data(test_sensor_raw, global_means, global_stds)
    # 4. Save metadata
    meta = {
        'means': global_means.tolist(),
        'stds': global_stds.tolist(),
        'input_shape': (1, train_sensor_raw[0].shape[0], train_sensor_raw[0].shape[1])
    }
    with open(META_SAVE_PATH, 'wb') as f:
        pickle.dump(meta, f)
    print(f"Saved global metadata to {META_SAVE_PATH}")
    # 5. K-Fold setup
    kf = KFold(n_splits=NUM_FOLDS, shuffle=True, random_state=SEED)
    fold_val_losses, fold_val_accs = [], []
    ensemble_models = []
    # 6. Cross-validation training
    for fold, (train_idx, val_idx) in enumerate(kf.split(train_sensor_normed)):
        print(f"\n{'='*50}")
        print(f"Starting fold {fold+1}/{NUM_FOLDS}")
        print(f"{'='*50}")
        X_train_fold = [train_sensor_normed[i] for i in train_idx]
        y_train_fold = [train_labels[i] for i in train_idx]
        X_val_fold = [train_sensor_normed[i] for i in val_idx]
        y_val_fold = [train_labels[i] for i in val_idx]
        model, history = train_model(
            X_train_fold, y_train_fold,
            X_val_fold, y_val_fold,
            batch_size=TRAIN_BATCH_SIZE,
            epochs=EPOCHS,
            lr=LEARNING_RATE,
            weight_decay=WEIGHT_DECAY
        )
        # Save model
        model_path = f"{MODEL_SAVE_PREFIX}fold{fold+1}.pth"
        torch.save(model.state_dict(), model_path)
        print(f"Saved model for fold {fold+1} to {model_path}")
        # Record validation performance
        val_loss = history['val_loss'][-1]
        val_acc = history['val_acc'][-1]
        fold_val_losses.append(val_loss)
        fold_val_accs.append(val_acc)
        print(f"Fold {fold+1} Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.4f}, Best Epoch: {history.get('best_epoch')}")
        # Plot history
        print(f"Plotting training history for fold {fold+1}...")
        plot_training_history(history, fold_idx=fold)
        # Evaluate on test set
        print(f"\nEvaluating fold {fold+1} model on test set...")
        evaluate_model(
            model,
            test_sensor_raw, test_labels,
            means=global_means, stds=global_stds,
            dataset_name=f"Test (Fold {fold+1})",
            batch_size=EVAL_BATCH_SIZE
        )
        ensemble_models.append(model)
    # 7. Average validation metrics
    avg_val_loss = np.mean(fold_val_losses)
    avg_val_acc = np.mean(fold_val_accs)
    print(f"\nAverage Validation Loss across {NUM_FOLDS} folds: {avg_val_loss:.4f}")
    print(f"Average Validation Accuracy across {NUM_FOLDS} folds: {avg_val_acc:.4f}")
    # 8. Soft voting ensemble
    print("\nEvaluating ensemble with soft voting on test set...")
    soft_voting_preds, soft_voting_probs = soft_voting_predict(
        ensemble_models, test_sensor_raw, global_means, global_stds,
        batch_size=EVAL_BATCH_SIZE
    )
    soft_accuracy = np.mean(soft_voting_preds == np.array(test_labels))
    print(f"\nTest (Soft Voting Ensemble) Accuracy: {soft_accuracy:.4f}")
    print(f"\n{classification_report(test_labels, soft_voting_preds, zero_division=0, target_names=['No Irregularity', 'Irregularity'])}")
    cm = confusion_matrix(test_labels, soft_voting_preds)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=['No Irregularity', 'Irregularity'], 
                yticklabels=['No Irregularity', 'Irregularity'])
    plt.title('Test (Soft Voting Ensemble) Confusion Matrix')
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.savefig('test_soft_voting_confusion_matrix.png')
    plt.show()
    # 9. Hard voting ensemble
    print("\nEvaluating ensemble with hard voting on test set...")
    hard_voting_preds = hard_voting_predict(
        ensemble_models, test_sensor_raw, global_means, global_stds,
        batch_size=EVAL_BATCH_SIZE
    )
    hard_accuracy = np.mean(hard_voting_preds == np.array(test_labels))
    print(f"\nTest (Hard Voting Ensemble) Accuracy: {hard_accuracy:.4f}")
    print(f"\n{classification_report(test_labels, hard_voting_preds, zero_division=0, target_names=['No Irregularity', 'Irregularity'])}")
    cm = confusion_matrix(test_labels, hard_voting_preds)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Oranges', 
                xticklabels=['No Irregularity', 'Irregularity'], 
                yticklabels=['No Irregularity', 'Irregularity'])
    plt.title('Test (Hard Voting Ensemble) Confusion Matrix')
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.savefig('test_hard_voting_confusion_matrix.png')
    plt.show()
    # 10. Compare voting strategies
    print("\nComparison of Voting Strategies:")
    print(f"Soft Voting Accuracy: {soft_accuracy:.4f}")
    print(f"Hard Voting Accuracy: {hard_accuracy:.4f}")
    print(f"Difference: {abs(soft_accuracy - hard_accuracy):.4f}")
    # 11. Agreement analysis
    agreement = np.mean(soft_voting_preds == hard_voting_preds)
    print(f"\nAgreement between soft and hard voting: {agreement:.4f}")
    disagreement_mask = soft_voting_preds != hard_voting_preds
    n_disagreements = np.sum(disagreement_mask)
    if n_disagreements > 0:
        print(f"\nFound {n_disagreements} samples with disagreement between voting methods:")
        print("Sample indices with disagreement:", np.where(disagreement_mask)[0])
        for idx in np.where(disagreement_mask)[0][:5]:
            print(f"\nAnalysis for sample {idx}:")
            print(f"True label: {test_labels[idx]}")
            print(f"Soft voting prediction: {soft_voting_preds[idx]}")
            print(f"Hard voting prediction: {hard_voting_preds[idx]}")
            model_preds = []
            for model in ensemble_models:
                sample = test_sensor_raw[idx:idx+1]
                normed_sample, _, _ = normalize_sensor_data(sample, global_means, global_stds)
                tensor_sample = torch.from_numpy(normed_sample[0].astype(np.float32)).unsqueeze(0).unsqueeze(0).to(device)
                model.eval()
                with torch.no_grad():
                    output = model(tensor_sample)
                    _, pred = torch.max(output, 1)
                    model_preds.append(pred.item())
            print(f"Individual model predictions: {model_preds}")
            print(f"Vote counts: Class 0: {model_preds.count(0)}, Class 1: {model_preds.count(1)}")
    print("\nTraining and evaluation with binary classification complete!")
```
